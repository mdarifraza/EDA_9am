```python
q1=np.percentile(pwage,25)
q2=np.percentile(pwage,50)
q3=np.percentile(pwage,75)
iqr=q3-q1
lb=q1-1.5*iqr
ub=q3+1.5*iqr
con1=pwage>ub
con2=pwage<lb
con=con1 | con2
outliers_df=visa_df[con]
outliers_df['prevailing_wage']
outliers_df['prevailing_wage'].values
outliers_list=outliers_df['prevailing_wage'].values.tolist()
outliers_list
```

```python
pwage_median=visa_df['prevailing_wage'].median()
pwage_data=visa_df['prevailing_wage']
new_wage_data=[]
for i in pwage_data:
    if i in outliers_list:
        new_wage_data.append(pwage_median)
    else:
        new_wage_data.append(i)
new_wage_data
```

## winsorization

- whatever the outliers less than lb fill with lb

- whatever the outliers greater than ub fill with ub

- in numpy we have method clip

```python
# step-1: get the wage data values
# step-2: calculate lb
# step-3: calculate ub
# step-4: checck the parameters np.clip()
# step-5: above visualization analysis
```

```python
new_pwage=visa_df['prevailing_wage']
new_pwage
```

```python
q1=np.percentile(new_pwage,25)
q2=np.percentile(new_pwage,50)
q3=np.percentile(new_pwage,75)
iqr=q3-q1
lb=q1-1.5*iqr
ub=q3+1.5*iqr
lb,ub
```

```python
np.clip(new_pwage,lb,ub)
```

```python
plt.figure(figsize=(12,5))
plt.subplot(2,2,1).hist(new_pwage)
plt.subplot(2,2,2).hist(new_pwage)
plt.subplot(2,2,3).boxplot(new_pwage,vert=False)
plt.subplot(2,2,4).boxplot(new_pwage,vert=False)
plt.savefig('k.png')
plt.show()
```
